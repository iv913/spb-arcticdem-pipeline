```python
#!/usr/bin/env python3
"""
Стадия A v2: построение авторитетного DEM для СПб в БС-77.

Изменения против v1:
  - STREAMING ATL08: скачиваем → парсим → удаляем по одной грануле.
    Диск в любой момент времени держит ≤1 гранулу ATL08 (~50 МБ)
    вместо всех 200-450 сразу. Решает OSError: No space left on device.
  - Тайловый apply_geoid: RAM-safe версия для больших растров СПб.
  - Опциональное сужение диапазона дат для экономии трафика/времени.
  - Ранняя QA-фильтрация по одной грануле: в память GDF попадают
    уже отфильтрованные точки, не сырые.

Цепочка:
    ArcticDEM → ICESat-2 (streaming) → plane fit → EGM2008 → БС-77

Предварительные условия:
    pip install earthaccess h5py pystac-client rioxarray rasterio \\
                geopandas shapely pyproj scipy scikit-learn numpy pandas osmnx
    projsync --user-writable-directory --file us_nga_egm08_25.tif
    export EARTHDATA_TOKEN="..."
"""

from __future__ import annotations

import gc
import json
import os
import sys
from dataclasses import dataclass, asdict
from pathlib import Path
from typing import Iterable

import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio
from shapely.geometry import Point
from pyproj import Transformer
from scipy import stats


# =============================================================================
#                                КОНФИГУРАЦИЯ
# =============================================================================

WORK_DIR = Path("./spb_dem")
TILES_DIR = WORK_DIR / "arcticdem_tiles"
ATL08_TMP = WORK_DIR / "atl08_tmp"   # временная папка, очищается по ходу
WORK_DIR.mkdir(parents=True, exist_ok=True)
TILES_DIR.mkdir(parents=True, exist_ok=True)
ATL08_TMP.mkdir(parents=True, exist_ok=True)

ARCTICDEM_TILES = ["32_73_1_1", "32_73_2_1"]
ARCTICDEM_COLLECTION = "arcticdem-mosaics-v4.1-2m"
PGC_STAC_URL = "https://stac.pgc.umn.edu/api/v1"

WORK_EPSG = 32636

DEM_VRT = WORK_DIR / "arcticdem.vrt"
DEM_ELL = WORK_DIR / "dem_ellipsoidal.tif"
DEM_COREG = WORK_DIR / "dem_coregistered.tif"
DEM_EGM = WORK_DIR / "dem_egm2008.tif"
DEM_BS77 = WORK_DIR / "dem_bs77.tif"

ATL08_GPKG = WORK_DIR / "atl08_ground.gpkg"
REPORT = WORK_DIR / "calibration_report.json"

AOI_BBOX_WGS84 = (29.5, 59.7, 30.7, 60.2)

# --- ICESat-2 -----------------------------------------------------------------
# Компромисс между покрытием и трафиком. Для калибровки плоского DEM
# 3 последних года данных абсолютно достаточно (миссия работает стабильно).
ATL08_VERSION = "007"                        # v006 retired (NSIDC), только v007
ATL08_DATE_RANGE = ("2022-01-01", "2025-12-31")  # 3 года хватит, streaming защищает диск
ATL08_KEEP_RAW = False                       # True = не удалять HDF5 после парсинга

# --- Паспортные GCP в БС-77 --------------------------------------------------
@dataclass
class GCP:
    name: str
    lon: float
    lat: float
    h_bs77: float
    source: str = "паспорт"


GCPS: list[GCP] = [
    GCP("Gorny_embankment",
        lon=30.269963, lat=59.929175, h_bs77=2.846,
        source="паспорт геодезической точки"),
    GCP("KZS_01_V1",
        lon=29.694830, lat=59.994314, h_bs77=4.347,
        source="паспорт метеоплощадки"),
    GCP("KZS_02_S1",
        lon=29.696649, lat=59.987700, h_bs77=3.367,
        source="паспорт метеоплощадки"),
    GCP("KZS_03_S2_north",
        lon=29.836873, lat=60.027940, h_bs77=3.770,
        source="паспорт метеоплощадки"),
]

# --- Параметры ATL08 фильтрации ---------------------------------------------
ATL08_MIN_PHOTONS = 20
ATL08_MAX_UNC_M = 1.0
ATL08_MAX_SLOPE_DEG = 5.0
ATL08_REMOVE_ON_BUILDINGS = True

# --- Параметры подгонки плоскости -------------------------------------------
COREG_USE_HUBER = True
COREG_TILT_SIGNIFICANT_M_PER_KM = 0.05
COREG_MAX_DZ_M = 10.0

# --- Тайлинг ----------------------------------------------------------------
GEOID_TILE_SIZE = 4096       # размер тайла для RAM-safe apply_geoid


# =============================================================================
#                          0. СКАЧИВАНИЕ ArcticDEM
# =============================================================================

def download_arcticdem() -> list[Path]:
    import pystac_client
    import rioxarray

    os.environ.setdefault("AWS_NO_SIGN_REQUEST", "YES")
    os.environ.setdefault("AWS_REGION", "us-west-2")

    full_ids = [f"{t}_2m_v4.1" for t in ARCTICDEM_TILES]
    print(f"ArcticDEM тайлы: {full_ids}")

    api = pystac_client.Client.open(PGC_STAC_URL)
    items = list(api.search(collections=[ARCTICDEM_COLLECTION], ids=full_ids).items())
    if not items:
        items = list(api.search(collections=[ARCTICDEM_COLLECTION],
                                query={"pgc:tile": {"in": ARCTICDEM_TILES}}).items())
    if not items:
        sys.exit(f"PGC STAC не нашёл тайлы {full_ids}")

    paths = []
    for item in items:
        fp = TILES_DIR / f"{item.id}.tif"
        if fp.exists():
            print(f"  {item.id}: уже скачан")
            paths.append(fp)
            continue
        asset_key = "dem" if "dem" in item.assets else "data"
        href = item.assets[asset_key].href
        print(f"  {item.id}: скачиваю...")
        da = rioxarray.open_rasterio(href)
        da.rio.to_raster(str(fp), compress="LZW", BIGTIFF="IF_SAFER")
        paths.append(fp)
    return paths


def merge_and_reproject(tiles: list[Path]) -> Path:
    from osgeo import gdal
    gdal.UseExceptions()

    if DEM_ELL.exists():
        print(f"{DEM_ELL}: уже существует, пропуск")
        return DEM_ELL

    print(f"Создаю VRT из {len(tiles)} тайлов...")
    gdal.BuildVRT(str(DEM_VRT), [str(t) for t in tiles],
                  options=gdal.BuildVRTOptions(resampleAlg="bilinear"))
    print(f"Reprojection в EPSG:{WORK_EPSG}, 2 м...")
    gdal.Warp(str(DEM_ELL), str(DEM_VRT), options=gdal.WarpOptions(
        format="GTiff", dstSRS=f"EPSG:{WORK_EPSG}",
        xRes=2, yRes=2, dstNodata=-9999, resampleAlg="bilinear",
        creationOptions=["COMPRESS=LZW", "TILED=YES", "BIGTIFF=YES"],
        multithread=True, warpMemoryLimit=512,
    ))
    print(f"  → {DEM_ELL}")
    return DEM_ELL


# =============================================================================
#           1. STREAMING: ATL08 гранула → скачать → парсить → удалить
# =============================================================================

BEAMS = ["gt1l", "gt1r", "gt2l", "gt2r", "gt3l", "gt3r"]


def _read_beam_filtered(h5, beam: str, bbox: tuple) -> pd.DataFrame | None:
    """Читаем ОДИН луч + сразу применяем QA+AOI фильтр."""
    grp_name = f"{beam}/land_segments"
    if grp_name not in h5:
        return None
    grp = h5[grp_name]

    def g(name):
        try:
            return grp[name][:]
        except KeyError:
            return None

    lat = g("latitude"); lon = g("longitude")
    h_te = g("terrain/h_te_best_fit")
    if lat is None or h_te is None:
        return None

    h_unc = g("terrain/h_te_uncertainty")
    n_phot = g("terrain/n_te_photons")
    slope = g("terrain/terrain_slope")
    msw = g("msw_flag"); cloud = g("cloud_flag_atm")
    night = g("night_flag")

    w, s, e, n = bbox
    slope_rad = slope if slope is not None else np.full_like(lat, np.nan)
    qa = (
        np.isfinite(lat) & np.isfinite(lon) & np.isfinite(h_te)
        & (lon >= w) & (lon <= e) & (lat >= s) & (lat <= n)
        & (h_te >= -20) & (h_te <= 200)
        & ((n_phot if n_phot is not None else 0) >= ATL08_MIN_PHOTONS)
        & ((h_unc if h_unc is not None else np.full_like(h_te, 999))
           <= ATL08_MAX_UNC_M)
        & ((msw if msw is not None else np.zeros_like(h_te)) == 0)
        & (np.isin(cloud if cloud is not None else np.zeros_like(h_te), [0, 1]))
        & (np.abs(slope_rad) <= np.deg2rad(ATL08_MAX_SLOPE_DEG))
    )
    if not np.any(qa):
        return None

    return pd.DataFrame({
        "lat": lat[qa], "lon": lon[qa], "h_te_ell": h_te[qa],
        "h_unc": (h_unc[qa] if h_unc is not None else np.nan),
        "n_phot": (n_phot[qa] if n_phot is not None else 0),
        "night": (night[qa] if night is not None else 0),
        "beam": beam,
    })


def stream_download_atl08(bbox: tuple, date_range: tuple) -> pd.DataFrame:
    """
    Streaming: ищем гранулы через earthaccess, скачиваем по одной,
    парсим, удаляем, переходим к следующей. Возвращаем общий DF.
    """
    token = os.environ.get("EARTHDATA_TOKEN", "")
    if not token:
        print("⚠ EARTHDATA_TOKEN не задан — ICESat-2 стадия будет пропущена.")
        return pd.DataFrame()

    try:
        import earthaccess
        import h5py
    except ImportError as e:
        print(f"⚠ {e} не установлен — стадия ICESat-2 пропущена.")
        return pd.DataFrame()

    auth = earthaccess.login(strategy="environment")
    if not auth.authenticated:
        print("⚠ earthaccess не авторизовался по токену.")
        return pd.DataFrame()

    print(f"Поиск ATL08 v{ATL08_VERSION} по AOI {bbox} за {date_range}...")
    results = earthaccess.search_data(
        short_name="ATL08", version=ATL08_VERSION,
        bounding_box=bbox, temporal=date_range,
    )
    print(f"  найдено гранул: {len(results)}")
    if not results:
        return pd.DataFrame()

    # Оценка трафика
    total_mb = 0
    try:
        for r in results:
            total_mb += float(r.size) if hasattr(r, "size") else 0
    except Exception:
        pass
    if total_mb:
        print(f"  суммарный объём: ~{total_mb:.0f} МБ "
              f"(streaming: на диске ≤ 100 МБ одновременно)")

    # --- Streaming loop ------------------------------------------------------
    frames = []
    n_total = len(results)
    for i, granule in enumerate(results, 1):
        tmp_files = []
        try:
            # earthaccess.download принимает список, но мы даём по одному
            downloaded = earthaccess.download(
                [granule], str(ATL08_TMP), show_progress=False)
            tmp_files = [Path(f) for f in downloaded if Path(f).exists()]
            if not tmp_files:
                continue
            fp = tmp_files[0]

            # Парсим сразу
            with h5py.File(fp, "r") as h5:
                beam_frames = []
                for beam in BEAMS:
                    df = _read_beam_filtered(h5, beam, bbox)
                    if df is not None and not df.empty:
                        beam_frames.append(df)
                if beam_frames:
                    granule_df = pd.concat(beam_frames, ignore_index=True)
                    granule_df["src"] = fp.name
                    frames.append(granule_df)

            # Прогресс
            if i % 10 == 0 or i == n_total:
                n_pts = sum(len(f) for f in frames)
                print(f"  [{i:4d}/{n_total}] обработано, накоплено "
                      f"{n_pts:,} отфильтрованных сегментов")

        except Exception as e:
            print(f"  ! гранула {i}: {e}")
        finally:
            # ВАЖНО: удаляем HDF5 после парсинга
            if not ATL08_KEEP_RAW:
                for fp in tmp_files:
                    try:
                        fp.unlink()
                    except OSError:
                        pass

    if not frames:
        print("⚠ Ни одной валидной точки не извлечено.")
        return pd.DataFrame()

    df = pd.concat(frames, ignore_index=True)
    print(f"Итог: {len(df):,} сегментов после streaming-фильтрации")
    return df


def apply_building_mask(df: pd.DataFrame) -> gpd.GeoDataFrame:
    """Отсечь точки, попавшие в OSM building-полигоны."""
    gdf = gpd.GeoDataFrame(
        df, geometry=gpd.points_from_xy(df["lon"], df["lat"]), crs=4326
    )
    if not ATL08_REMOVE_ON_BUILDINGS:
        return gdf
    try:
        import osmnx as ox
        ox.settings.timeout = 180
        print("OSM building-полигоны...")
        bld = ox.features_from_bbox(bbox=AOI_BBOX_WGS84, tags={"building": True})
        bld = bld[bld.geometry.type.isin(["Polygon", "MultiPolygon"])]
        if not bld.empty:
            before = len(gdf)
            gdf = gpd.sjoin(gdf, bld[["geometry"]].to_crs(4326),
                            how="left", predicate="intersects")
            gdf = gdf[gdf["index_right"].isna()].drop(columns=["index_right"])
            print(f"  удалено на зданиях: {before - len(gdf):,}")
    except Exception as e:
        print(f"  ! OSM building-маска не применена: {e}")
    return gdf.reset_index(drop=True)


# =============================================================================
#                   2. САМПЛИРОВАНИЕ + ПОДГОНКА ПЛОСКОСТИ
# =============================================================================

def sample_dem_at(dem_path: Path, gdf_wgs: gpd.GeoDataFrame) -> gpd.GeoDataFrame:
    """Выборка DEM в каждой точке ICESat-2."""
    with rasterio.open(dem_path) as src:
        gdf_m = gdf_wgs.to_crs(src.crs)
        xy = [(p.x, p.y) for p in gdf_m.geometry]
        vals = np.array([v[0] for v in src.sample(xy)], dtype=float)
        nodata = src.nodata if src.nodata is not None else -9999
        vals[vals == nodata] = np.nan
        vals[vals < -1000] = np.nan

    gdf = gdf_wgs.copy()
    gdf["h_dem_ell"] = vals
    gdf["dZ"] = gdf["h_te_ell"] - gdf["h_dem_ell"]
    gdf_m = gdf.to_crs(WORK_EPSG)
    gdf["x_m"] = gdf_m.geometry.x
    gdf["y_m"] = gdf_m.geometry.y
    before = len(gdf)
    gdf = gdf[np.isfinite(gdf["dZ"]) & (gdf["dZ"].abs() < COREG_MAX_DZ_M)].copy()
    print(f"  После |dZ|<{COREG_MAX_DZ_M} м: {len(gdf):,}/{before:,}")
    return gdf


def fit_plane(gdf: gpd.GeoDataFrame) -> tuple[dict, np.ndarray]:
    x = gdf["x_m"].to_numpy()
    y = gdf["y_m"].to_numpy()
    z = gdf["dZ"].to_numpy()
    x0, y0 = float(np.median(x)), float(np.median(y))
    X = np.column_stack([x - x0, y - y0])

    if COREG_USE_HUBER:
        from sklearn.linear_model import HuberRegressor
        model = HuberRegressor(epsilon=1.35, max_iter=200).fit(X, z)
        a, b, c = float(model.intercept_), *map(float, model.coef_)
    else:
        A = np.column_stack([np.ones_like(x), x - x0, y - y0])
        coef, *_ = np.linalg.lstsq(A, z, rcond=None)
        a, b, c = map(float, coef)

    tilt_m_per_km = 1000 * float(np.hypot(b, c))
    use_tilt = tilt_m_per_km >= COREG_TILT_SIGNIFICANT_M_PER_KM
    if not use_tilt:
        a = float(np.median(z))
        b = c = 0.0

    pred = a + b * X[:, 0] + c * X[:, 1]
    resid = z - pred

    summary = {
        "n_points": int(len(z)),
        "x0_m": x0, "y0_m": y0,
        "offset_a_m": a, "tilt_b_per_m": b, "tilt_c_per_m": c,
        "tilt_m_per_km": tilt_m_per_km,
        "tilt_used": bool(use_tilt),
        "dZ_before_median_m": float(np.median(z)),
        "dZ_before_mad_m": float(stats.median_abs_deviation(z, scale="normal")),
        "dZ_after_median_m": float(np.median(resid)),
        "dZ_after_mad_m": float(stats.median_abs_deviation(resid, scale="normal")),
    }
    print(f"  Плоскость: a={a:+.3f} м, тилт={tilt_m_per_km:.3f} м/км  "
          f"{'(применён)' if use_tilt else '(отвергнут)'}")
    print(f"  MAD dZ: {summary['dZ_before_mad_m']:.3f} → "
          f"{summary['dZ_after_mad_m']:.3f} м")
    return summary, np.array([a, b, c, x0, y0])


# =============================================================================
#                  3. ПРИМЕНЕНИЕ ПЛОСКОСТИ (тайлово)
# =============================================================================

def apply_plane(dem_in: Path, dem_out: Path, plane: np.ndarray) -> None:
    a, b, c, x0, y0 = plane
    with rasterio.open(dem_in) as src:
        profile = src.profile.copy()
        profile.update(dtype="float32", compress="LZW", predictor=3,
                       BIGTIFF="IF_SAFER")
        tr_to_work = Transformer.from_crs(src.crs, WORK_EPSG, always_xy=True)
        with rasterio.open(dem_out, "w", **profile) as dst:
            for _, window in src.block_windows(1):
                data = src.read(1, window=window).astype(np.float32)
                rows, cols = np.mgrid[
                    window.row_off:window.row_off + window.height,
                    window.col_off:window.col_off + window.width]
                xs, ys = rasterio.transform.xy(
                    src.transform, rows.ravel(), cols.ravel())
                xm, ym = tr_to_work.transform(np.asarray(xs), np.asarray(ys))
                shift = (a + b * (xm.reshape(data.shape) - x0)
                         + c * (ym.reshape(data.shape) - y0)).astype(np.float32)
                out = data + shift
                if src.nodata is not None:
                    out[data == src.nodata] = src.nodata
                dst.write(out, 1, window=window)
    print(f"  → {dem_out}")


# =============================================================================
#              4. ГЕОИД EGM2008 (RAM-safe тайловая версия)
# =============================================================================

def apply_geoid(dem_in: Path, dem_out: Path,
                tile_sz: int = GEOID_TILE_SIZE) -> None:
    """
    Тайловое вычитание геоида: H_orth = H_ell - N(lat,lon).
    RAM-safe для больших растров (20k×20k пикселей и больше).
    """
    with rasterio.open(dem_in) as src:
        profile = src.profile.copy()
        profile.update(dtype="float32", compress="LZW", predictor=3,
                       BIGTIFF="IF_SAFER")
        nodata = src.nodata
        rows, cols = src.height, src.width
        tr_to_wgs = Transformer.from_crs(src.crs, 4326, always_xy=True)
        try:
            pipe = Transformer.from_pipeline(
                "+proj=pipeline "
                "+step +proj=axisswap +order=2,1 "
                "+step +proj=unitconvert +xy_in=deg +xy_out=rad "
                "+step +proj=vgridshift +grids=us_nga_egm08_25.tif +multiplier=1"
            )
        except Exception as e:
            sys.exit(f"Ошибка PROJ (нужна сетка us_nga_egm08_25.tif): {e}\n"
                     f"Установите: projsync --user-writable-directory "
                     f"--file us_nga_egm08_25.tif")

        with rasterio.open(dem_out, "w", **profile) as dst:
            n_tiles = ((rows + tile_sz - 1) // tile_sz) \
                      * ((cols + tile_sz - 1) // tile_sz)
            t_idx = 0
            for r0 in range(0, rows, tile_sz):
                for c0 in range(0, cols, tile_sz):
                    t_idx += 1
                    r1 = min(r0 + tile_sz, rows)
                    c1 = min(c0 + tile_sz, cols)
                    window = rasterio.windows.Window(c0, r0, c1 - c0, r1 - r0)
                    data = src.read(1, window=window).astype(np.float64)
                    valid = (data != nodata) if nodata is not None \
                        else ~np.isnan(data)
                    if not np.any(valid):
                        out = data.astype(np.float32)
                        if nodata is not None:
                            out[~valid] = nodata
                        dst.write(out, 1, window=window)
                        continue
                    rr, cc = np.mgrid[r0:r1, c0:c1]
                    xs, ys = rasterio.transform.xy(
                        src.transform, rr.ravel(), cc.ravel())
                    xs = np.asarray(xs, dtype=np.float64)
                    ys = np.asarray(ys, dtype=np.float64)
                    lons, lats = tr_to_wgs.transform(xs, ys)
                    _, _, minus_N = pipe.transform(
                        lons, lats, np.zeros_like(lons))
                    N = -np.asarray(minus_N, dtype=np.float32).reshape(data.shape)
                    out = (data - N).astype(np.float32)
                    if nodata is not None:
                        out[~valid] = nodata
                    dst.write(out, 1, window=window)
                    if t_idx % max(1, n_tiles // 5) == 0 or t_idx == n_tiles:
                        print(f"    Геоид: тайл {t_idx}/{n_tiles} "
                              f"({100*t_idx/n_tiles:.0f}%)")
    print(f"  → {dem_out}")


# =============================================================================
#          5. ПАСПОРТНЫЕ ТОЧКИ → ФИНАЛЬНЫЙ СДВИГ EGM2008 → БС-77
# =============================================================================

def _sample_bilinear(dem_path: Path, lon: float, lat: float) -> float | None:
    with rasterio.open(dem_path) as src:
        tr = Transformer.from_crs(4326, src.crs, always_xy=True)
        x, y = tr.transform(lon, lat)
        col = (x - src.transform.c) / src.transform.a
        row = (y - src.transform.f) / src.transform.e
        c0, r0 = int(np.floor(col)), int(np.floor(row))
        dx, dy = col - c0, row - r0
        if not (0 <= r0 < src.height - 1 and 0 <= c0 < src.width - 1):
            return None
        win = rasterio.windows.Window(c0, r0, 2, 2)
        tile = src.read(1, window=win).astype(float)
        if src.nodata is not None:
            tile[tile == src.nodata] = np.nan
        if np.any(np.isnan(tile)):
            return None
        return float(
            (1 - dx) * (1 - dy) * tile[0, 0]
            + dx * (1 - dy) * tile[0, 1]
            + (1 - dx) * dy * tile[1, 0]
            + dx * dy * tile[1, 1]
        )


def compute_bs77_offset(dem_egm_path: Path,
                        gcps: list[GCP]) -> tuple[float, dict]:
    per_point = []
    for g in gcps:
        v = _sample_bilinear(dem_egm_path, g.lon, g.lat)
        if v is None:
            print(f"  ! {g.name}: вне растра или NoData")
            continue
        delta = g.h_bs77 - v
        per_point.append({
            "name": g.name, "lon": g.lon, "lat": g.lat,
            "h_passport_bs77": g.h_bs77, "h_dem_egm": round(v, 3),
            "delta_m": round(delta, 3), "source": g.source,
        })
        print(f"  {g.name:20s}  DEM_EGM={v:+.3f}  паспорт={g.h_bs77:+.3f}  "
              f"Δ={delta:+.3f} м")

    if not per_point:
        sys.exit("Ни одна паспортная точка не попала в растр.")

    deltas = np.array([r["delta_m"] for r in per_point])
    offset = float(np.median(deltas))
    mad = float(stats.median_abs_deviation(deltas, scale="normal")) \
        if len(deltas) > 1 else 0.0

    loo = []
    for i in range(len(deltas)):
        rest = np.delete(deltas, i)
        pred = float(np.median(rest)) if len(rest) else 0.0
        loo.append({
            "name": per_point[i]["name"],
            "predicted_m": round(pred, 3),
            "actual_m": round(float(deltas[i]), 3),
            "residual_m": round(float(deltas[i] - pred), 3),
        })

    rms_loo = float(np.sqrt(np.mean([r["residual_m"] ** 2 for r in loo]))) \
        if loo else 0.0

    print(f"\n  Медианный сдвиг EGM→БС-77: {offset:+.3f} м (MAD={mad:.3f})")
    print(f"  LOO RMS: {rms_loo:.3f} м")

    return offset, {
        "egm2008_to_bs77_offset_m": offset, "n_gcps": len(per_point),
        "mad_m": mad, "loo_rms_m": rms_loo,
        "per_point": per_point, "loo_residuals": loo,
    }


def apply_constant_offset(dem_in: Path, dem_out: Path, offset: float) -> None:
    with rasterio.open(dem_in) as src:
        profile = src.profile.copy()
        profile.update(dtype="float32", compress="LZW", predictor=3,
                       BIGTIFF="IF_SAFER")
        with rasterio.open(dem_out, "w", **profile) as dst:
            for _, window in src.block_windows(1):
                data = src.read(1, window=window).astype(np.float32)
                out = data + np.float32(offset)
                if src.nodata is not None:
                    out[data == src.nodata] = src.nodata
                dst.write(out, 1, window=window)
    print(f"  → {dem_out}")


# =============================================================================
#                                  ГЛАВНЫЙ
# =============================================================================

def main() -> None:
    report = {"stages": {}, "config": {
        "arcticdem_tiles": ARCTICDEM_TILES,
        "atl08_version": ATL08_VERSION,
        "atl08_date_range": list(ATL08_DATE_RANGE),
        "epsg": WORK_EPSG,
        "aoi_bbox_wgs84": list(AOI_BBOX_WGS84),
        "gcps": [asdict(g) for g in GCPS],
    }}

    print("\n" + "=" * 70 + "\n[0/5] ArcticDEM\n" + "=" * 70)
    tiles = download_arcticdem()
    dem_ell = merge_and_reproject(tiles)

    print("\n" + "=" * 70 + "\n[1/5] ICESat-2 ATL08 (streaming)\n" + "=" * 70)
    df = stream_download_atl08(AOI_BBOX_WGS84, ATL08_DATE_RANGE)
    gc.collect()

    if not df.empty and len(df) >= 30:
        gdf = apply_building_mask(df)
        del df; gc.collect()
        gdf = sample_dem_at(dem_ell, gdf)
        gdf.to_file(ATL08_GPKG, driver="GPKG", layer="atl08_with_dem")
        summary, plane = fit_plane(gdf)
        report["stages"]["icesat2_coreg"] = summary
        print("\n" + "=" * 70 + "\n[2/5] Применение плоскости к DEM\n" + "=" * 70)
        apply_plane(dem_ell, DEM_COREG, plane)
        dem_coreg = DEM_COREG
    else:
        print("Коррегистрация пропущена")
        dem_coreg = dem_ell
        report["stages"]["icesat2_coreg"] = {"skipped": True, "n_points": len(df)}

    print("\n" + "=" * 70 + "\n[3/5] Геоид EGM2008 (tiled)\n" + "=" * 70)
    apply_geoid(dem_coreg, DEM_EGM)

    print("\n" + "=" * 70 + "\n[4/5] 4 паспортные точки → БС-77\n" + "=" * 70)
    offset, bs77_report = compute_bs77_offset(DEM_EGM, GCPS)
    report["stages"]["bs77_anchor"] = bs77_report
    apply_constant_offset(DEM_EGM, DEM_BS77, offset)

    print("\n" + "=" * 70 + "\n[5/5] Отчёт\n" + "=" * 70)
    report["final_dem"] = str(DEM_BS77)
    REPORT.write_text(json.dumps(report, indent=2, ensure_ascii=False))
    print(f"\n✓ Готово. DEM: {DEM_BS77}")
    print(f"  Отчёт: {REPORT}")

    # Очистка временной папки ATL08
    if not ATL08_KEEP_RAW:
        for f in ATL08_TMP.glob("*"):
            try: f.unlink()
            except OSError: pass


if __name__ == "__main__":
    main()
```
